```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset_path = './Big_QRs/train'
maps = np.load(f'{dataset_path}/maps.npy', mmap_mode='c')
targets = np.load(f'{dataset_path}/cf.npy', mmap_mode='c')
```

```python
from tqdm.notebook import tqdm
```

```python
traps = []
for i in tqdm(range(len(maps))):
    traps.append(np.sum(maps[i, 0]))
```

```python
plt.hist(traps, bins=100)
plt.show()
```

```python
mn = np.min(traps)
mx = np.max(traps)
print(mn, mx)
```

```python
for i in range(len(maps)):
    if np.sum(maps[i, 0]) == mn:
        plt.imshow(1 - maps[i, 0], cmap='gray')
        plt.show()
        plt.imshow(targets[i, 0])
        plt.show()
```

```python
for i in range(len(maps)):
    if np.sum(maps[i, 0]) == mx:
        plt.imshow(1 - maps[i, 0], cmap='gray')
        plt.show()
        plt.imshow(targets[i, 0])
        plt.show()
```

```python
med = np.median(traps)
print(np.median(traps))
```

```python
for i in range(len(maps)):
    if np.sum(maps[i, 0]) == med:
        plt.imshow(1 - maps[i, 0], cmap='gray')
        plt.show()
        plt.imshow(targets[i, 0])
        plt.show()
```
