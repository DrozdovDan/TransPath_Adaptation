```python
import os
import numpy as np
from skimage.filters import threshold_otsu
from PIL import Image
import glob
from natsort import natsorted


input_dir = "/home/silvarum/planning-datasets/data/mpd/original"

maze_types = [d for d in os.listdir(input_dir) 
                 if os.path.isdir(os.path.join(input_dir, d))]

def process_mazes(output_dir, maze_size=32):
    
    for maze_type in maze_types:
        test_path = os.path.join(input_dir, maze_type, "test")
        image_paths = natsorted(glob.glob(os.path.join(test_path, "*.png")))
        if not image_paths:
            continue
        
        processed_images = []
        for image_path in image_paths:
            image = np.asarray(
                Image.open(image_path).convert("L").resize((maze_size, maze_size)),
                dtype=np.float32,
            )
            th = threshold_otsu(image)
            image_out = np.zeros_like(image)
            image_out[image > th] = 1.0
            processed_images.append(image_out)
    
        if not os.path.exists(output_dir):
            os.makedirs(output_dir)
                
        if processed_images:
            tensor = np.array(processed_images)
            output_file = os.path.join(output_dir, f"{maze_type}")
            np.save(output_file, tensor)
    print("smth")
    return True

process_mazes("/home/silvarum/TransPath_Adaptation/results/32mp")
```
